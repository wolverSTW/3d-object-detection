# 02 - KITTI Data Audit

This notebook audits the raw KITTI dataset and confirms the files required for training are available.

Goals:
- verify raw files exist
- summarize dataset layout
- inspect labels and calibration files
- prepare the data for preprocessing

```python
from pathlib import Path

ROOT = Path.cwd().resolve().parent
DATA_DIR = ROOT / 'data' / 'KITTI' / 'raw'

required = ['image_2', 'label_2', 'calib']
for folder in required:
    path = DATA_DIR / folder
    print(f'{folder}:', 'OK' if path.exists() else 'MISSING')
    if path.exists():
        files = sorted(path.iterdir())[:5]
        print(' sample:', [f.name for f in files])
```

```python
# Check label file contents and first few samples
label_dir = DATA_DIR / 'label_2'
if label_dir.exists():
    files = sorted(label_dir.glob('*.txt'))
    print('num label files:', len(files))
    for f in files[:2]:
        print('\nFILE:', f.name)
        with f.open('r', encoding='utf-8') as fh:
            for i, line in enumerate(fh):
                print(line.strip())
                if i >= 1:
                    break
else:
    print('label_2 directory is missing.')
```

```python
# Check a few calibration files
calib_dir = DATA_DIR / 'calib'
if calib_dir.exists():
    files = sorted(calib_dir.glob('*.txt'))
    print('num calib files:', len(files))
    for f in files[:2]:
        print('\nCALIB:', f.name)
        with f.open('r', encoding='utf-8') as fh:
            for i, line in enumerate(fh):
                print(line.strip())
                if i >= 2:
                    break
else:
    print('calib directory is missing.')
```

## Audit summary

A healthy dataset should contain:
- `image_2` with RGB input images
- `label_2` with KITTI annotation files
- `calib` with calibration matrices

If these folders are missing, the server setup step must be repeated before training begins.
