# 05 - KITTI EDA and Class Distribution

This notebook performs a lightweight exploratory data analysis on KITTI labels and image metadata.

Objectives:
- inspect label class frequencies
- summarize sample counts
- identify annotation quality issues
- prepare the analysis for model design and ablation decisions

```python
from pathlib import Path
import collections

ROOT = Path.cwd().resolve().parent
RAW = ROOT / 'data' / 'KITTI' / 'raw'
LABEL_DIR = RAW / 'label_2'

if LABEL_DIR.exists():
    files = sorted(LABEL_DIR.glob('*.txt'))
    print('label files:', len(files))
    counts = collections.Counter()
    valid = 0
    for f in files:
        with f.open('r', encoding='utf-8') as fh:
            for line in fh:
                parts = line.strip().split()
                if len(parts) < 15:
                    continue
                counts[parts[0]] += 1
                valid += 1
    print('valid label rows:', valid)
    print('class counts:', dict(counts))
else:
    print('label_2 directory not found. Please download or unzip KITTI first.')
```

```python
# Image and calibration counts
image_dir = RAW / 'image_2'
calib_dir = RAW / 'calib'

print('image files:', len(list(image_dir.glob('*.png'))) if image_dir.exists() else 0)
print('calib files:', len(list(calib_dir.glob('*.txt'))) if calib_dir.exists() else 0)
```

```python
# Display a few sample KITTI labels from the dataset
if LABEL_DIR.exists():
    files = sorted(LABEL_DIR.glob('*.txt'))
    for f in files[:2]:
        print('\nFILE:', f.name)
        with f.open('r', encoding='utf-8') as fh:
            for i, line in enumerate(fh):
                print(line.strip())
                if i >= 2:
                    break
else:
    print('No label files found.')
```

## EDA interpretation

The class distribution reveals dataset imbalance, which matters for model calibration and recall. In KITTI, cars usually dominate the distribution, while cyclists and pedestrians are sparser and more difficult to localize.

This analysis guides the training strategy, class weighting, and evaluation interpretation for the thesis.
