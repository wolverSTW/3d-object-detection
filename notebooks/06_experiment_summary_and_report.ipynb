# 06 - Experiment Summary and Reporting

This notebook formalizes the experiment reporting workflow for the thesis.

It is intended to:
- track experiment outputs
- compare different runs
- summarize metrics and efficiency
- prepare results for final analysis

```python
from pathlib import Path

ROOT = Path.cwd().resolve().parent
EXPERIMENTS = ROOT / 'experiments'
OUTPUTS = ROOT / 'outputs'

for d in [EXPERIMENTS, OUTPUTS]:
    print(d.name, 'exists:', d.exists())
    if d.exists():
        print(' contents:', [p.name for p in sorted(d.iterdir())[:10]])
```

```python
# A compact experiment summary template
summary = {
    'dataset': 'KITTI',
    'task': 'Monocular 3D Object Detection',
    'models': ['baseline', 'geometry-guided'],
    'metrics': ['AP3D', 'APBEV', 'Distance MAE', 'FPS'],
    'report_dir': 'outputs',
    'notes': 'Compare baseline vs geometry-guided model across Easy/Moderate/Hard splits.'
}

print(summary)
```
