This notebook connects the checked dataset with the experimental pipeline.

Planned workflow:
1. validate KITTI dataset structure
2. load the training split
3. train baseline or geometry model
4. evaluate outputs with KITTI-style metrics
5. record results in the outputs folder

```python
from pathlib import Path

ROOT = Path.cwd().resolve().parent
DATA_ROOT = ROOT / 'data' / 'KITTI'
RAW = DATA_ROOT / 'raw'
SPLIT = DATA_ROOT / 'splits' / 'train.txt'

print('DATA_ROOT:', DATA_ROOT)
print('RAW exists:', RAW.exists())
print('SPLIT exists:', SPLIT.exists())
```

```python
# Optional: create a split if the image folder is available
image_dir = RAW / 'image_2'
if image_dir.exists():
    images = sorted(image_dir.glob('*.png'))
    if images:
        ids = [p.stem for p in images]
        SPLIT.parent.mkdir(parents=True, exist_ok=True)
        SPLIT.write_text('\n'.join(ids) + '\n', encoding='utf-8')
        print(f'Prepared split for {len(ids)} images')
```

```python
# Example training command for a real GPU server
cmd = (
    'python scripts/train.py '
    '--config configs/experiments/baseline.yaml '
    '--model baseline '
    '--epochs 50'
)
print(cmd)
```
