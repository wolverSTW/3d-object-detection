# 01 - KITTI Download and Setup

This notebook prepares the KITTI dataset for the thesis project workflow.

Objectives:
- verify the dataset folder structure
- check raw label/calibration files
- prepare a basic train split
- document the setup for server use

```python
from pathlib import Path
import os

ROOT = Path.cwd().resolve().parent
DATA_DIR = ROOT / 'data' / 'KITTI' / 'raw'
SPLIT_DIR = ROOT / 'data' / 'KITTI' / 'splits'

print(f'Project root: {ROOT}')
print(f'Data dir: {DATA_DIR}')
print(f'Split dir: {SPLIT_DIR}')

for p in [DATA_DIR, SPLIT_DIR]:
    p.mkdir(parents=True, exist_ok=True)
    print(f'Ensured folder exists: {p}')
```

```python
# Check for expected KITTI subfolders
expected = ['image_2', 'label_2', 'calib']
for name in expected:
    path = DATA_DIR / name
    print(f'{name}: exists={path.exists()}')
    if path.exists():
        print(' sample files:', len(list(path.iterdir())[:5]))
```

```python
# Create a basic train split if image files are present
img_dir = DATA_DIR / 'image_2'
if img_dir.exists():
    ids = sorted(p.stem for p in img_dir.glob('*.png'))
    if ids:
        SPLIT_DIR.mkdir(parents=True, exist_ok=True)
        train_path = SPLIT_DIR / 'train.txt'
        with train_path.open('w', encoding='utf-8') as f:
            for idx in ids:
                f.write(f'{idx}\n')
        print(f'Wrote {len(ids)} sample IDs to {train_path}')
    else:
        print('No PNG files found in image_2.')
else:
    print('image_2 directory not found. Download the KITTI dataset first.')
```

## Notes

This notebook is intentionally lightweight. On a real server, you would download the official KITTI dataset into `data/KITTI/raw` before proceeding to EDA and preprocessing.
